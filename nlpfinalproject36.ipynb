```python
from typing import List
from pydantic import BaseModel, Field


class WordAnalysis(BaseModel):

    word: str = Field(
        description="Original word from the input text"
    )

    language: str = Field(
        description=(
            "Language of the word. "
            "Use English, Gujarati, Hindi, or Other."
        )
    )

    normalized: str = Field(
        description="English meaning of the word"
    )


class CodeMixAnalysis(BaseModel):

    original_text: str = Field(
        description="The original user input"
    )

    detected_languages: List[str] = Field(
        description="Languages detected in the text"
    )

    language_distribution: dict[str, float] = Field(
        description=(
            "Approximate percentage of each language"
        )
    )

    code_mixing_level: str = Field(
        description=(
            "Low, Medium, or High"
        )
    )

    switching_points: int = Field(
        description=(
            "Approximate number of language switches"
        )
    )

    word_analysis: List[WordAnalysis] = Field(
        description="Word-level language analysis"
    )

    normalized_text: str = Field(
        description=(
            "Meaning of the complete text in English"
        )
    )

    sentiment: str = Field(
        description=(
            "Positive, Negative, Neutral, or Mixed"
        )
    )

    intent: str = Field(
        description="Main intent of the user"
    )

    emotion: str = Field(
        description="Main emotion expressed"
    )

    explanation: str = Field(
        description="Short explanation of the analysis"
    )


print("CodeMixAI data structure created successfully!")
```

```python
def calculate_cmis(
    language_distribution,
    switching_points,
    total_words,
    transliterated_words
):
    """
    Calculate the proposed Code-Mixing Intelligence Score.

    Score range: 0 - 100
    """

    if total_words <= 0:
        return 0.0

    # ------------------------------------------------
    # 1. Language diversity: 0-30
    # ------------------------------------------------

    active_languages = len([
        language
        for language, percentage
        in language_distribution.items()
        if percentage > 0
    ])

    diversity_score = min(
        active_languages / 3,
        1
    ) * 30

    # ------------------------------------------------
    # 2. Switching frequency: 0-30
    # ------------------------------------------------

    possible_switches = max(
        total_words - 1,
        1
    )

    switching_ratio = (
        switching_points /
        possible_switches
    )

    switching_score = min(
        switching_ratio,
        1
    ) * 30

    # ------------------------------------------------
    # 3. Transliteration: 0-20
    # ------------------------------------------------

    transliteration_ratio = (
        transliterated_words /
        total_words
    )

    transliteration_score = min(
        transliteration_ratio,
        1
    ) * 20

    # ------------------------------------------------
    # 4. Language balance: 0-20
    # ------------------------------------------------

    percentages = [
        percentage
        for percentage
        in language_distribution.values()
        if percentage > 0
    ]

    balance_score = 0

    if len(percentages) >= 2:

        maximum = max(percentages)
        minimum = min(percentages)

        if maximum > 0:

            balance_score = (
                minimum / maximum
            ) * 20

    # ------------------------------------------------
    # Final CMIS
    # ------------------------------------------------

    score = (
        diversity_score
        + switching_score
        + transliteration_score
        + balance_score
    )

    return round(
        min(score, 100),
        2
    )


print("CMIS engine created successfully!")
```

```python
def detect_transliterated_words(text):

    words = text.lower().split()

    transliterated = []

    for word in words:

        # Remove simple punctuation
        clean_word = (
            word
            .strip(".,!?;:'\"()[]{}")
        )

        if (
            clean_word in GUJARATI_ROMAN_WORDS
            or
            clean_word in HINDI_ROMAN_WORDS
        ):

            transliterated.append(
                clean_word
            )

    return transliterated


transliterated_words = (
    detect_transliterated_words(
        test_text
    )
)

print(
    "Transliterated words:"
)

print(
    transliterated_words
)
```

```python
def interpret_cmis(score):

    if score < 25:

        return "Low code-mixing"

    elif score < 50:

        return "Moderate code-mixing"

    elif score < 75:

        return "High code-mixing"

    else:

        return "Very high code-mixing"


cmis_level = interpret_cmis(
    cmis_score
)

print(
    "CMIS Interpretation:",
    cmis_level
)
```

```python
def complete_analysis(text):

    # 1. Get LLM analysis
    result = analyze_code_mixed_text(text)

    # 2. Count words
    total_words = count_words(
        result.original_text
    )

    # 3. Detect Romanized Gujarati/Hindi words
    transliterated = detect_transliterated_words(
        result.original_text
    )

    # 4. Calculate CMIS
    cmis_score = calculate_cmis(
        language_distribution=result.language_distribution,
        switching_points=result.switching_points,
        total_words=total_words,
        transliterated_words=len(transliterated)
    )

    # 5. Interpret score
    cmis_level = interpret_cmis(
        cmis_score
    )

    return {
        "analysis": result,
        "total_words": total_words,
        "transliterated_words": transliterated,
        "cmis_score": cmis_score,
        "cmis_level": cmis_level
    }


print("Complete CodeMixAI analysis pipeline created successfully!")
```

```python
# STEP 28 — Local Code-Mixing Score

def calculate_local_mixing_score(text):

    words = text.lower().split()

    if not words:
        return 0

    gujarati_count = 0
    hindi_count = 0
    english_count = 0

    for word in words:

        clean_word = word.strip(".,!?;:'\"()[]{}")

        if clean_word in GUJARATI_ROMAN_WORDS:
            gujarati_count += 1

        elif clean_word in HINDI_ROMAN_WORDS:
            hindi_count += 1

        else:
            english_count += 1

    total = len(words)

    non_english = gujarati_count + hindi_count

    score = (non_english / total) * 100

    return round(score, 2)


for item in test_dataset:

    score = calculate_local_mixing_score(item["text"])

    print("\nText:", item["text"])
    print("Local mixing score:", score, "%")
```

```python
# STEP 29 — Local Language Distribution

def local_language_distribution(text):

    words = text.lower().split()

    counts = {
        "English": 0,
        "Gujarati": 0,
        "Hindi": 0
    }

    for word in words:

        clean_word = word.strip(".,!?;:'\"()[]{}")

        if clean_word in GUJARATI_ROMAN_WORDS:
            counts["Gujarati"] += 1

        elif clean_word in HINDI_ROMAN_WORDS:
            counts["Hindi"] += 1

        else:
            counts["English"] += 1

    total = sum(counts.values())

    if total == 0:
        return counts

    distribution = {
        language: round((count / total) * 100, 2)
        for language, count in counts.items()
    }

    return distribution


for item in test_dataset:

    print("\nText:", item["text"])
    print(local_language_distribution(item["text"]))
```

```python
# STEP 30 — Language Switching Detection

def detect_language(word):

    clean_word = word.lower().strip(".,!?;:'\"()[]{}")

    if clean_word in GUJARATI_ROMAN_WORDS:
        return "Gujarati"

    elif clean_word in HINDI_ROMAN_WORDS:
        return "Hindi"

    else:
        return "English"


def detect_switching_points(text):

    words = text.split()

    if len(words) < 2:
        return []

    switches = []

    previous_language = detect_language(words[0])

    for i in range(1, len(words)):

        current_language = detect_language(words[i])

        if current_language != previous_language:

            switches.append({
                "position": i,
                "from": previous_language,
                "to": current_language,
                "word": words[i]
            })

        previous_language = current_language

    return switches


for item in test_dataset:

    switches = detect_switching_points(item["text"])

    print("\nText:", item["text"])
    print("Switching points:", len(switches))
    print(switches)
```

```python
# STEP 31 — Local CodeMix Analyzer

def local_analysis(text):

    statistics = local_text_statistics(text)

    distribution = local_language_distribution(text)

    transliterated = detect_transliterated_words(text)

    switches = detect_switching_points(text)

    mixing_score = calculate_local_mixing_score(text)

    return {
        "original_text": text,
        "statistics": statistics,
        "language_distribution": distribution,
        "transliterated_words": transliterated,
        "switching_points": switches,
        "switching_count": len(switches),
        "local_mixing_score": mixing_score
    }


local_result = local_analysis(
    "Aaje project complete karvu che but I don't know how to start."
)

print(local_result)
```

```python
# STEP 42.3 — Robust CodeMixAI Chat Function

from langchain_core.messages import HumanMessage, AIMessage

chat_history = []


def get_response_text(response):
    """
    Convert different LangChain/Gemini response formats
    into a normal string.
    """

    content = response.content

    # Normal string response
    if isinstance(content, str):
        return content

    # Gemini may return a list of content blocks
    if isinstance(content, list):

        text_parts = []

        for item in content:

            if isinstance(item, str):
                text_parts.append(item)

            elif isinstance(item, dict):

                if "text" in item:
                    text_parts.append(
                        str(item["text"])
                    )

        return "\n".join(text_parts)

    return str(content)


def codemix_chat(user_message):

    global chat_history

    if not user_message or not user_message.strip():
        return "Please enter a message."

    try:

        messages = chat_prompt.format_messages(
            history=chat_history,
            input=user_message
        )

        response = chat_llm.invoke(messages)

        assistant_message = get_response_text(response)

        # Save conversation
        chat_history.append(
            HumanMessage(
                content=user_message
            )
        )

        chat_history.append(
            AIMessage(
                content=assistant_message
            )
        )

        return assistant_message

    except Exception as e:

        return (
            "⚠️ CodeMixAI could not generate a response.\n\n"
            f"Error: {type(e).__name__}\n"
            f"Details: {str(e)}"
        )


print("CodeMixAI chat function created successfully!")
```

```python
%%writefile codemix_engine.py

import re


GUJARATI_WORDS = {
    "aaje", "aavse", "aavya", "aavyu",
    "che", "chhe", "chu", "hato", "hati",
    "bahu", "saru", "saras", "nathi",
    "mane", "mari", "mara", "maru",
    "tame", "tamne", "tamaro", "tamari",
    "shu", "su", "kem", "kya",
    "ane", "pan", "ma", "thi", "mate",
    "karvu", "karvanu", "karo", "kari",
    "javu", "javanu", "jaavu",
    "game", "gamtu", "gamy",
    "thayu", "thase", "hase", "hoy",
    "karyu", "kare", "karta",
    "karine", "levu", "levanu",
    "aavvu", "aavanu",
    "che", "chhe"
}


HINDI_WORDS = {
    "mujhe", "mujhko", "mujh",
    "mera", "meri", "mere",
    "hum", "hamara",
    "tum", "tumhara", "tumhari",
    "aap", "aapka", "aapki",
    "kya", "kyun", "kyunki",
    "nahi", "nahin",
    "hai", "hain",
    "tha", "thi", "the",
    "bahut", "accha", "achha",
    "aaj", "kal", "kaise",
    "karna", "karni", "karo",
    "jaana", "jana", "jao",
    "raha", "rahi", "rahe",
    "chahiye", "lagta", "lagti",
    "ho", "hona", "hoga",
    "ka", "ki", "ke",
    "ko", "se", "par",
    "liye", "phir", "abhi",
    "sab", "kuch"
}


def clean_word(word):

    return word.lower().strip(
        ".,!?;:'\"()[]{}<>|/\\"
    )


def detect_language(word):

    word = clean_word(word)

    if word in GUJARATI_WORDS:
        return "Gujarati"

    if word in HINDI_WORDS:
        return "Hindi"

    return "English"


def analyze_text(text):

    if not text or not text.strip():
        return {
            "error": "Empty text"
        }

    words = text.split()

    languages = []
    transliterated = []

    counts = {
        "English": 0,
        "Gujarati": 0,
        "Hindi": 0
    }

    for word in words:

        language = detect_language(word)

        languages.append(language)

        counts[language] += 1

        if language != "English":
            transliterated.append(
                clean_word(word)
            )

    total_words = len(words)

    # --------------------------------
    # DISTRIBUTION
    # --------------------------------

    distribution = {}

    for language in counts:

        distribution[language] = round(
            counts[language] / total_words * 100,
            2
        )

    # --------------------------------
    # SWITCHES
    # --------------------------------

    switches = []

    for i in range(1, len(languages)):

        if languages[i] != languages[i - 1]:

            switches.append({
                "position": i + 1,
                "from": languages[i - 1],
                "to": languages[i],
                "word": words[i]
            })

    # --------------------------------
    # SWITCHING DENSITY
    # --------------------------------

    switching_density = round(
        len(switches) / max(total_words - 1, 1),
        3
    )

    # --------------------------------
    # LEXICAL DIVERSITY
    # --------------------------------

    cleaned_words = [
        clean_word(word)
        for word in words
    ]

    unique_words = len(
        set(cleaned_words)
    )

    lexical_diversity = round(
        unique_words / total_words,
        3
    )

    # --------------------------------
    # LANGUAGE DOMINANCE
    # --------------------------------

    dominant_language = max(
        counts,
        key=counts.get
    )

    # --------------------------------
    # MIXING SCORE
    # --------------------------------

    non_english = (
        counts["Gujarati"] +
        counts["Hindi"]
    )

    mixing_score = round(
        non_english / total_words * 100,
        2
    )

    # --------------------------------
    # CMIS
    # --------------------------------

    active_languages = sum(
        1
        for value in counts.values()
        if value > 0
    )

    diversity_score = (
        active_languages / 3
    ) * 30

    switching_score = min(
        switching_density,
        1
    ) * 30

    transliteration_score = min(
        len(transliterated) / total_words,
        1
    ) * 20

    percentages = [
        value
        for value in distribution.values()
        if value > 0
    ]

    balance_score = 0

    if len(percentages) >= 2:

        maximum = max(percentages)
        minimum = min(percentages)

        if maximum > 0:

            balance_score = (
                minimum / maximum
            ) * 20

    cmis = round(
        min(
            diversity_score +
            switching_score +
            transliteration_score +
            balance_score,
            100
        ),
        2
    )

    # --------------------------------
    # CMIS LEVEL
    # --------------------------------

    if cmis < 25:
        level = "Low"

    elif cmis < 50:
        level = "Moderate"

    elif cmis < 75:
        level = "High"

    else:
        level = "Very High"

    return {

        "text": text,

        "words": words,

        "total_words": total_words,

        "unique_words": unique_words,

        "distribution": distribution,

        "language_sequence": languages,

        "transliterated_words": transliterated,

        "switches": switches,

        "switching_count": len(switches),

        "switching_density":
            switching_density,

        "lexical_diversity":
            lexical_diversity,

        "dominant_language":
            dominant_language,

        "mixing_score":
            mixing_score,

        "cmis":
            cmis,

        "cmis_level":
            level
    }
```

```python
%%writefile ai_engine.py

import os

try:

    from langchain_google_genai import (
        ChatGoogleGenerativeAI
    )

    from langchain_core.messages import (
        SystemMessage,
        HumanMessage
    )

except Exception:

    ChatGoogleGenerativeAI = None
    SystemMessage = None
    HumanMessage = None


SYSTEM_PROMPT = """
You are CodeMixAI, a conversational AI system
specialized in code-mixed Indian languages.

You understand:

- Gujarati-English
- Hindi-English
- Hinglish
- Romanized Gujarati
- Romanized Hindi
- Gujarati-Hindi-English

Understand the meaning of code-mixed sentences
without assuming that every Romanized word is English.

Respond naturally.

For technical questions provide practical
examples and explanations.

Do not unnecessarily translate the user's message.
"""


def create_model():

    api_key = os.environ.get(
        "GOOGLE_API_KEY"
    )

    if not api_key:

        return None

    if ChatGoogleGenerativeAI is None:

        return None

    try:

        return ChatGoogleGenerativeAI(
            model="gemini-3.8-flash",
            google_api_key=api_key,
            temperature=0.7
        )

    except Exception:

        return None


def generate_response(
    user_message,
    history=None
):

    model = create_model()

    if model is None:

        return {
            "success": False,
            "response":
                "Gemini is not currently available."
        }

    messages = [
        SystemMessage(
            content=SYSTEM_PROMPT
        )
    ]

    if history:

        for item in history:

            if item["role"] == "user":

                messages.append(
                    HumanMessage(
                        content=item["content"]
                    )
                )

    messages.append(
        HumanMessage(
            content=user_message
        )
    )

    try:

        result = model.invoke(
            messages
        )

        content = result.content

        if isinstance(content, str):

            response = content

        else:

            response = str(content)

        return {
            "success": True,
            "response": response
        }

    except Exception as e:

        error_text = str(e)

        if (
            "429" in error_text
            or
            "RESOURCE_EXHAUSTED"
            in error_text
            or
            "quota" in error_text.lower()
        ):

            return {
                "success": False,
                "response":
                    "⚠️ Gemini API quota is "
                    "currently exhausted. "
                    "The local CodeMixAI analyzer "
                    "is still available."
            }

        return {
            "success": False,
            "response":
                "⚠️ AI service error: "
                + type(e).__name__
        }
```

```python
%%writefile smart_engine.py

from codemix_engine import analyze_text
from ai_engine import generate_response


def smart_analyze(
    text,
    history=None
):

    analysis = analyze_text(text)

    if "error" in analysis:

        return {
            "analysis": analysis,
            "ai_response":
                "Please enter some text."
        }

    ai_result = generate_response(
        text,
        history
    )

    return {

        "analysis": analysis,

        "ai_response":
            ai_result["response"],

        "ai_available":
            ai_result["success"]
    }
```

```python
%%writefile security_engine.py

MAX_TEXT_LENGTH = 5000


def validate_input(text):

    if text is None:

        return {
            "valid": False,
            "message":
                "Input cannot be empty."
        }

    text = text.strip()

    if not text:

        return {
            "valid": False,
            "message":
                "Please enter some text."
        }

    if len(text) > MAX_TEXT_LENGTH:

        return {
            "valid": False,
            "message":
                f"Text is too long. "
                f"Maximum allowed length is "
                f"{MAX_TEXT_LENGTH} characters."
        }

    return {
        "valid": True,
        "message":
            "Input is valid."
    }
```

```python
%%writefile error_handler.py

def safe_error_message(error):

    error_text = str(error)

    if "429" in error_text:

        return (
            "Gemini quota exceeded. "
            "Please try again after the quota resets."
        )

    if "401" in error_text:

        return (
            "Authentication failed. "
            "Please check your API key."
        )

    if "403" in error_text:

        return (
            "API access was denied. "
            "Please verify your Gemini API configuration."
        )

    if "timeout" in error_text.lower():

        return (
            "The AI service timed out. "
            "Please try again."
        )

    return (
        "An unexpected error occurred: "
        + type(error).__name__
    )
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd

from codemix_engine import analyze_text
from smart_engine import smart_analyze
from export_engine import analysis_to_json, analysis_to_csv
from security_engine import validate_input


# ============================================================
# PAGE CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="CodeMixAI",
    page_icon="🌐",
    layout="wide"
)


# ============================================================
# SESSION STATE
# ============================================================

if "messages" not in st.session_state:
    st.session_state.messages = []

if "history" not in st.session_state:
    st.session_state.history = []


# ============================================================
# SIDEBAR
# ============================================================

st.sidebar.title("🌐 CodeMixAI")

st.sidebar.caption(
    "Code-Mixed Language Understanding Platform"
)

mode = st.sidebar.radio(
    "Navigation",
    [
        "🔍 Analyzer",
        "🤖 AI Assistant",
        "🧠 Smart Mode",
        "📊 Analytics"
    ]
)

st.sidebar.markdown("---")

st.sidebar.info(
    """
Supported Languages:

• Gujarati-English
• Hindi-English
• Hinglish
• Romanized Gujarati
• Romanized Hindi
"""
)


# ============================================================
# HEADER
# ============================================================

st.title("🌐 CodeMixAI")

st.caption(
    "NLP + LLM + Code-Mixed Language Intelligence"
)


# ============================================================
# ANALYZER MODE
# ============================================================

if mode == "🔍 Analyzer":

    st.header("🔍 Code-Mix Analyzer")

    text = st.text_area(
        "Enter code-mixed text",
        height=160,
        placeholder=(
            "Example: Aaje mara Python project ma "
            "error aave che."
        )
    )

    analyze_button = st.button(
        "🚀 Analyze Text",
        use_container_width=True
    )

    if analyze_button:

        validation = validate_input(text)

        if not validation["valid"]:

            st.error(validation["message"])

        else:

            result = analyze_text(text)

            if "error" in result:

                st.error(result["error"])

            else:

                # Save result
                st.session_state.history.append(result)

                st.success(
                    "Analysis completed successfully."
                )

                # ----------------------------------------
                # METRICS
                # ----------------------------------------

                col1, col2, col3, col4 = st.columns(4)

                col1.metric(
                    "Total Words",
                    result["total_words"]
                )

                col2.metric(
                    "Unique Words",
                    result["unique_words"]
                )

                col3.metric(
                    "Language Switches",
                    result["switching_count"]
                )

                col4.metric(
                    "CMIS",
                    result["cmis"]
                )

                st.markdown("---")

                # ----------------------------------------
                # LANGUAGE DISTRIBUTION
                # ----------------------------------------

                st.subheader(
                    "🌍 Language Distribution"
                )

                distribution_df = pd.DataFrame(
                    {
                        "Language": list(
                            result["distribution"].keys()
                        ),
                        "Percentage": list(
                            result["distribution"].values()
                        )
                    }
                )

                st.bar_chart(
                    distribution_df.set_index(
                        "Language"
                    )
                )

                # ----------------------------------------
                # CMIS
                # ----------------------------------------

                st.subheader(
                    "🧠 Code-Mixing Intelligence Score"
                )

                st.progress(
                    min(int(result["cmis"]), 100)
                )

                st.write(
                    f"CMIS: **{result['cmis']}/100**"
                )

                st.write(
                    f"Level: **{result['cmis_level']}**"
                )

                # ----------------------------------------
                # ADDITIONAL METRICS
                # ----------------------------------------

                c1, c2, c3 = st.columns(3)

                c1.metric(
                    "Mixing Score",
                    f"{result['mixing_score']}%"
                )

                c2.metric(
                    "Switching Density",
                    result["switching_density"]
                )

                c3.metric(
                    "Lexical Diversity",
                    result["lexical_diversity"]
                )

                # ----------------------------------------
                # DOMINANT LANGUAGE
                # ----------------------------------------

                st.subheader(
                    "🏆 Dominant Language"
                )

                st.info(
                    result["dominant_language"]
                )

                # ----------------------------------------
                # TRANSLITERATION
                # ----------------------------------------

                st.subheader(
                    "🔤 Transliterated Words"
                )

                transliterated = result[
                    "transliterated_words"
                ]

                if transliterated:

                    st.write(
                        ", ".join(transliterated)
                    )

                else:

                    st.write(
                        "No transliterated words detected."
                    )

                # ----------------------------------------
                # LANGUAGE SWITCHING
                # ----------------------------------------

                st.subheader(
                    "🔄 Language Switching Points"
                )

                switches = result["switches"]

                if switches:

                    switch_df = pd.DataFrame(
                        switches
                    )

                    st.dataframe(
                        switch_df,
                        use_container_width=True
                    )

                else:

                    st.info(
                        "No language switching detected."
                    )

                # ----------------------------------------
                # WORD LEVEL ANALYSIS
                # ----------------------------------------

                st.subheader(
                    "📝 Word-Level Language Analysis"
                )

                word_rows = []

                for word, language in zip(
                    result["words"],
                    result["language_sequence"]
                ):

                    word_rows.append(
                        {
                            "Word": word,
                            "Detected Language": language
                        }
                    )

                word_df = pd.DataFrame(
                    word_rows
                )

                st.dataframe(
                    word_df,
                    use_container_width=True
                )

                # ----------------------------------------
                # EXPORT
                # ----------------------------------------

                st.subheader(
                    "📥 Export"
                )

                json_data = analysis_to_json(
                    result
                )

                csv_data = analysis_to_csv(
                    result
                )

                col1, col2 = st.columns(2)

                with col1:

                    st.download_button(
                        "⬇️ Download JSON",
                        data=json_data,
                        file_name="codemix_analysis.json",
                        mime="application/json"
                    )

                with col2:

                    st.download_button(
                        "⬇️ Download CSV",
                        data=csv_data,
                        file_name="codemix_analysis.csv",
                        mime="text/csv"
                    )


# ============================================================
# AI ASSISTANT MODE
# ============================================================

elif mode == "🤖 AI Assistant":

    st.header(
        "🤖 CodeMixAI Assistant"
    )

    st.write(
        "Chat with CodeMixAI using English, "
        "Hindi, Gujarati or mixed language."
    )

    # ----------------------------------------
    # DISPLAY OLD MESSAGES
    # ----------------------------------------

    for message in st.session_state.messages:

        with st.chat_message(
            message["role"]
        ):

            st.markdown(
                message["content"]
            )

    # ----------------------------------------
    # CHAT INPUT
    # ----------------------------------------

    user_input = st.chat_input(
        "Type your message..."
    )

    if user_input:

        # Save user message
        st.session_state.messages.append(
            {
                "role": "user",
                "content": user_input
            }
        )

        with st.chat_message("user"):

            st.markdown(user_input)

        # ----------------------------------------
        # LOCAL ANALYSIS
        # ----------------------------------------

        local_result = analyze_text(
            user_input
        )

        # ----------------------------------------
        # AI RESPONSE
        # ----------------------------------------

        smart_result = smart_analyze(
            user_input,
            st.session_state.messages
        )

        response = smart_result[
            "ai_response"
        ]

        ai_available = smart_result[
            "ai_available"
        ]

        # ----------------------------------------
        # FALLBACK
        # ----------------------------------------

        if not ai_available:

            detected_languages = []

            for language, percentage in (
                local_result["distribution"].items()
            ):

                if percentage > 0:

                    detected_languages.append(
                        language
                    )

            language_text = ", ".join(
                detected_languages
            )

            response = (
                response
                + "\n\n"
                + "**Local Code-Mix Analysis:**\n"
                + f"- Languages: {language_text}\n"
                + f"- CMIS: {local_result['cmis']}/100\n"
                + f"- Mixing Score: "
                f"{local_result['mixing_score']}%\n"
                + f"- Language Switches: "
                f"{local_result['switching_count']}"
            )

        # ----------------------------------------
        # DISPLAY RESPONSE
        # ----------------------------------------

        with st.chat_message("assistant"):

            st.markdown(response)

        # Save response
        st.session_state.messages.append(
            {
                "role": "assistant",
                "content": response
            }
        )

    # ----------------------------------------
    # CLEAR CHAT
    # ----------------------------------------

    if st.button(
        "🗑️ Clear Conversation"
    ):

        st.session_state.messages = []

        st.rerun()


# ============================================================
# SMART MODE
# ============================================================

elif mode == "🧠 Smart Mode":

    st.header(
        "🧠 Smart Code-Mix Intelligence"
    )

    st.write(
        """
Smart Mode combines the local NLP engine
with the generative AI engine.
"""
    )

    text = st.text_area(
        "Enter your code-mixed text",
        height=160,
        placeholder=(
            "Example: Mujhe Python project banana hai "
            "but I don't know how to start."
        )
    )

    if st.button(
        "🚀 Run Smart Analysis",
        use_container_width=True
    ):

        validation = validate_input(text)

        if not validation["valid"]:

            st.error(
                validation["message"]
            )

        else:

            result = smart_analyze(
                text,
                st.session_state.messages
            )

            analysis = result["analysis"]

            # ----------------------------------------
            # METRICS
            # ----------------------------------------

            c1, c2, c3, c4 = st.columns(4)

            c1.metric(
                "CMIS",
                analysis["cmis"]
            )

            c2.metric(
                "Mixing Score",
                f"{analysis['mixing_score']}%"
            )

            c3.metric(
                "Switches",
                analysis["switching_count"]
            )

            c4.metric(
                "Dominant Language",
                analysis["dominant_language"]
            )

            # ----------------------------------------
            # LANGUAGE DISTRIBUTION
            # ----------------------------------------

            st.subheader(
                "🌍 Language Distribution"
            )

            distribution_df = pd.DataFrame(
                {
                    "Language":
                        list(
                            analysis[
                                "distribution"
                            ].keys()
                        ),

                    "Percentage":
                        list(
                            analysis[
                                "distribution"
                            ].values()
                        )
                }
            )

            st.bar_chart(
                distribution_df.set_index(
                    "Language"
                )
            )

            # ----------------------------------------
            # AI RESPONSE
            # ----------------------------------------

            st.subheader(
                "🤖 AI Interpretation"
            )

            st.write(
                result["ai_response"]
            )

            # ----------------------------------------
            # SWITCHES
            # ----------------------------------------

            st.subheader(
                "🔄 Language Switching"
            )

            if analysis["switches"]:

                st.dataframe(
                    pd.DataFrame(
                        analysis["switches"]
                    ),
                    use_container_width=True
                )

            else:

                st.info(
                    "No switching points detected."
                )


# ============================================================
# ANALYTICS MODE
# ============================================================

elif mode == "📊 Analytics":

    st.header(
        "📊 CodeMixAI Analytics"
    )

    history = st.session_state.history

    if not history:

        st.info(
            "No analysis history yet."
        )

        st.write(
            "Go to **Analyzer** and analyze "
            "some text first."
        )

    else:

        rows = []

        for index, item in enumerate(
            history,
            start=1
        ):

            rows.append(
                {
                    "Analysis": index,
                    "Words":
                        item["total_words"],
                    "Unique Words":
                        item["unique_words"],
                    "Switches":
                        item["switching_count"],
                    "Mixing Score":
                        item["mixing_score"],
                    "CMIS":
                        item["cmis"],
                    "Dominant Language":
                        item["dominant_language"]
                }
            )

        analytics_df = pd.DataFrame(
            rows
        )

        # ----------------------------------------
        # SUMMARY
        # ----------------------------------------

        c1, c2, c3 = st.columns(3)

        c1.metric(
            "Total Analyses",
            len(analytics_df)
        )

        c2.metric(
            "Average CMIS",
            round(
                analytics_df["CMIS"].mean(),
                2
            )
        )

        c3.metric(
            "Average Mixing Score",
            round(
                analytics_df["Mixing Score"].mean(),
                2
            )
        )

        # ----------------------------------------
        # CMIS GRAPH
        # ----------------------------------------

        st.subheader(
            "📈 CMIS Trend"
        )

        st.line_chart(
            analytics_df.set_index(
                "Analysis"
            )["CMIS"]
        )

        # ----------------------------------------
        # MIXING SCORE GRAPH
        # ----------------------------------------

        st.subheader(
            "🔀 Mixing Score"
        )

        st.bar_chart(
            analytics_df.set_index(
                "Analysis"
            )["Mixing Score"]
        )

        # ----------------------------------------
        # TABLE
        # ----------------------------------------

        st.subheader(
            "📋 Analysis History"
        )

        st.dataframe(
            analytics_df,
            use_container_width=True
        )

        # ----------------------------------------
        # EXPORT
        # ----------------------------------------

        st.download_button(
            "⬇️ Download Analytics CSV",
            data=analytics_df.to_csv(
                index=False
            ),
            file_name="codemixai_analytics.csv",
            mime="text/csv"
        )


# ============================================================
# FOOTER
# ============================================================

st.sidebar.markdown("---")

st.sidebar.caption(
    "CodeMixAI • NLP + LLM + LangChain"
)

st.sidebar.caption(
    "Academic / Research Prototype"
)
```

```python
from pathlib import Path

files = {}

files["normalization_engine.py"] = r'''
GUJARATI_NORMALIZATION = {
    "aaje": "today",
    "mane": "I",
    "mara": "my",
    "mari": "my",
    "maru": "my",
    "tame": "you",
    "tamne": "you",
    "tamaro": "your",
    "tamari": "your",
    "che": "is",
    "chhe": "is",
    "chu": "am",
    "bahu": "very",
    "saru": "good",
    "saras": "good",
    "nathi": "not",
    "ane": "and",
    "pan": "but",
    "ma": "in",
    "thi": "from",
    "mate": "for",
    "kem": "why",
    "shu": "what",
    "karvu": "to do",
    "karvanu": "to do",
    "karo": "do",
    "kari": "did",
    "javu": "to go",
    "javanu": "to go",
    "game": "like",
    "thayu": "happened",
    "thase": "will happen"
}

HINDI_NORMALIZATION = {
    "mujhe": "I",
    "mujhko": "I",
    "mujh": "me",
    "mera": "my",
    "meri": "my",
    "mere": "my",
    "tum": "you",
    "tumhara": "your",
    "tumhari": "your",
    "aap": "you",
    "aapka": "your",
    "aapki": "your",
    "kya": "what",
    "kyun": "why",
    "kyunki": "because",
    "nahi": "not",
    "nahin": "not",
    "hai": "is",
    "hain": "are",
    "bahut": "very",
    "accha": "good",
    "achha": "good",
    "aaj": "today",
    "kal": "tomorrow",
    "kaise": "how",
    "karna": "to do",
    "karni": "to do",
    "karo": "do",
    "jaana": "go",
    "jana": "go",
    "jao": "go",
    "raha": "doing",
    "rahi": "doing",
    "rahe": "doing",
    "chahiye": "need",
    "lagta": "seems",
    "lagti": "seems",
    "ka": "of",
    "ki": "of",
    "ke": "of",
    "ko": "to",
    "se": "from",
    "par": "on",
    "liye": "for",
    "phir": "then",
    "abhi": "now"
}

def normalize_text(text):
    words = text.split()
    normalized = []

    for word in words:
        clean = word.lower().strip(
            ".,!?;:'\"()[]{}<>|/\\"
        )

        if clean in GUJARATI_NORMALIZATION:
            normalized.append(GUJARATI_NORMALIZATION[clean])
        elif clean in HINDI_NORMALIZATION:
            normalized.append(HINDI_NORMALIZATION[clean])
        else:
            normalized.append(clean)

    return " ".join(normalized)
'''

files["semantic_engine.py"] = r'''
POSITIVE_WORDS = {
    "good", "great", "awesome", "nice",
    "excellent", "happy", "love", "like",
    "saras", "accha", "achha", "mast"
}

NEGATIVE_WORDS = {
    "bad", "boring", "sad", "angry", "hate",
    "problem", "error", "fail", "nathi", "bored"
}

QUESTION_WORDS = {
    "what", "why", "how", "when", "where",
    "kya", "kyun", "kaise", "shu", "kem"
}

def detect_sentiment(text):
    words = text.lower().split()

    positive = 0
    negative = 0

    for word in words:
        clean = word.strip(
            ".,!?;:'\"()[]{}"
        )

        if clean in POSITIVE_WORDS:
            positive += 1

        if clean in NEGATIVE_WORDS:
            negative += 1

    if positive > negative:
        return "Positive"

    if negative > positive:
        return "Negative"

    return "Neutral"


def detect_intent(text):
    words = text.lower().split()

    cleaned = {
        word.strip(
            ".,!?;:'\"()[]{}"
        )
        for word in words
    }

    if cleaned.intersection(QUESTION_WORDS):
        return "Question"

    if "help" in cleaned:
        return "Help Request"

    if "error" in cleaned:
        return "Technical Problem"

    if "project" in cleaned:
        return "Project Related"

    return "General Conversation"
'''

files["memory_engine.py"] = r'''
MAX_MESSAGES = 20

def add_message(messages, role, content):
    messages.append({
        "role": role,
        "content": content
    })

    if len(messages) > MAX_MESSAGES:
        del messages[0:len(messages) - MAX_MESSAGES]

    return messages


def clear_memory():
    return []
'''

files["chat_export.py"] = r'''
import json

def export_chat(messages):
    return json.dumps(
        messages,
        indent=4,
        ensure_ascii=False
    )
'''

files["logger_engine.py"] = r'''
import logging

logging.basicConfig(
    filename="codemixai.log",
    level=logging.INFO,
    format=(
        "%(asctime)s - "
        "%(levelname)s - "
        "%(message)s"
    )
)

def log_analysis(text):
    logging.info(
        "Text analyzed. Length=%d",
        len(text)
    )


def log_error(error):
    logging.error(
        "Application error: %s",
        str(error)
    )
'''

for filename, content in files.items():
    Path(filename).write_text(
        content.strip() + "\n",
        encoding="utf-8"
    )
    print("✅ Restored:", filename)

print("\n🎉 All five missing modules have been restored.")
```

```python
from codemix_engine import analyze_text
from evaluation_dataset import EVALUATION_DATASET

correct = 0
total = len(EVALUATION_DATASET)

for item in EVALUATION_DATASET:
    result = analyze_text(item["text"])

    detected = set(
        language
        for language, percentage
        in result["distribution"].items()
        if percentage > 0
    )

    expected = set(item["expected_languages"])

    if detected == expected:
        correct += 1

    print("ID:", item["id"])
    print("Expected:", expected)
    print("Detected:", detected)
    print("-" * 50)

accuracy = correct / total * 100

print("\n==============================")
print("Evaluation Accuracy:", round(accuracy, 2), "%")
print("==============================")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
from codemix_engine import detect_language

y_true = []
y_pred = []

for item in EVALUATION_DATASET:

    text = item["text"]
    words = text.split()

    expected_languages = item["expected_languages"]

    for word in words:
        predicted = detect_language(word)

        if predicted in expected_languages:
            actual = predicted
        else:
            actual = expected_languages[0]

        y_true.append(actual)
        y_pred.append(predicted)

precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1 Score:", round(f1, 4))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

labels = ["English", "Gujarati", "Hindi"]

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=labels
)

print("Confusion Matrix:")
print(cm)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

display.plot()
plt.title("Code-Mixed Language Detection Confusion Matrix")
plt.show()
```
